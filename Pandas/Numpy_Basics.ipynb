```python
import numpy as np
# 1D aaray
a = np.array([1,2,3,4])

# 2D array
b = np.array([[1,2,3,4],[5,6,7,8]])

# 3D array
c = np.array([[[1,2],[3,4],[5,6],[7,8]]])

print("1D: ",a)
print("2D: ",b)
print("3D: ",c)
```

```python
# challenge 1
# Imagine you're analysing the examination results of 5 students across 4 subjects.
```

```python
import numpy as np
marks = np.array([
    [78, 85, 90, 92],
    [82, 88, 91, 87],
    [76, 82, 88, 90],
    [92, 95, 98, 94],
    [88, 92, 96, 90]
])
print(marks)
```

```python
print("Shape: ",marks.shape)
print("Dimensions: ",marks.ndim)
print("Size: ",marks.size)
```

```python
stu_avg = np.mean(marks, axis = 1)
print(stu_avg)
```

```python
subject_avg = np.mean(marks, axis=0)

print(subject_avg)
```

```python
print("Highest mark:", np.max(marks))
print("Lowest mark:", np.min(marks))
```

```python
high_performers = stu_avg[stu_avg > 80]

print(high_performers)
```

```python
topper_index = np.argmax(stu_avg)

print("Topper: Student", topper_index + 1)
print("Average:", stu_avg[topper_index])
```

```python
print("Student averages:", stu_avg)
print("Subject averages:", subject_avg)
print("Topper: Student", topper_index + 1)
print("Students scoring above 80:", len(high_performers))
```
