```python
data = {
    'Name': ['Rahul', 'Priya', 'Aman', 'Neha'],
    'Age': [25, 28, 22, 30],
    'Salary': [30000, 45000, 25000, 55000],
    'Purchased': ['Yes', 'No', 'Yes', 'Yes']
}

df = pd.DataFrame(data)
print(df)
```
